```python
import socket
import json

SOCKET_PATH = "/tmp/fridge_bus.sock"

client = socket.socket(socket.AF_UNIX, socket.SOCK_STREAM)
client.connect(SOCKET_PATH)

# session_id = "test_one_apple"
session_id = "test_multi_put"
# session_id = "test_put_then_take" # Short Period Take
message = {
    "type": "SESSION_CLOSED",
    "session_id": session_id
}

client.sendall(json.dumps(message).encode("utf-8"))
client.close()

print("Sent session_id: ", session_id)
```
